Given a dataset, match treated (X equals 1) to untreated (X equals 0) based on the confounder (Z).  

Find the average treatment effect (each item corresponds to one counterfactual) where the counterfactual is the nearest item in the other group (you can use Nearest Neighbors for this.) 

Find the average treatment effect on the treated, where each treated item corresponds to a counterfactual untreated item, but we otherwise ignore the untreated items. 

Find the average treatment effect on the untreated, where each untreated item corresponds to a counterfactual treated item, but we otherwise ignore the treated items. 

Find the optimal treatment effect, which is the maximum treatment effect across all untreated items (i.e., it ends up considering only a single untreated item with its single counterfactual). 
Use the file homework_6.1.csv. 

```python
# import necessary libraries
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.neighbors import NearestNeighbors
```

```python
df= pd.read_csv("../data/homework_6.1.csv")
df.head()
```

## Question 1
Which is closest to the average treatment effect? 

- A: 1.832

- B: 2.014

- C: 1.583

- D: 1.695

```python
treated = df[df['X'] == 1].copy()
untreated = df[df['X'] == 0].copy()

# For each treated row, find nearest untreated row using Z
nn_untreated = NearestNeighbors(n_neighbors=1)
nn_untreated.fit(untreated[['Z']])

_, idx_u = nn_untreated.kneighbors(treated[['Z']])

treated_effects = (treated['Y'].to_numpy()- untreated.iloc[idx_u.flatten()]['Y'].to_numpy())

# For each untreated row, find nearest treated row using Z
nn_treated = NearestNeighbors(n_neighbors=1)
nn_treated.fit(treated[['Z']])

distances, idx_t = nn_treated.kneighbors(untreated[['Z']])

untreated_effects = (treated.iloc[idx_t.flatten()]['Y'].to_numpy()- untreated['Y'].to_numpy())

# ATE = average effect over ALL rows
all_effects = np.concatenate([treated_effects, untreated_effects])

ate = all_effects.mean()

ate
```

Answer: D
